```python
import yaml

yaml_file_path = "test.yaml"

# Open the YAML file and load its contents
with open(yaml_file_path, 'r') as yaml_file:
    data = yaml.safe_load(yaml_file)


print(data)
print(type(data))
```

```python
data["name"]
```

```python
data["email"]
```

```python
data.name
```

YAML files load into Python as dictionaries

ConfigBox is a small helper class from the python-box library.
It converts a dictionary into an object-like structure.

```python
from box import ConfigBox
```

```python
d = ConfigBox(data)
```
