## Ensure Annotation

```python
def get_product(x:int, y:int):
    return x*y
```

```python
get_product(x=2,y=4)
```

```python
get_product(x=2,y="4")
```

### Above we can see that instead of throwing a error get_product multiplied string 2 times 
### with the help of ensure annotations errors will be displayed

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_product(x:int, y:int):
     return x*y
```

```python
get_product(x=2,y="4")
```
